```python
import pandas as pd

df=pd.read_csv('customer_shopping_behavior.csv')
df.head()
```

```python
df.info()
```

```python
df.describe(include='all')
```

```python
df.isnull().sum()
```

```python
df['Review Rating']=df.groupby('Category')['Review Rating'].transform(lambda x: x.fillna(x.median()))
```

```python
df.isnull().sum()
```

```python
df.columns=df.columns.str.lower()
df.columns=df.columns.str.replace(" ","_")
df=df.rename(columns={'purchase amount (usd)':'purchase amount'})
df.columns
```

```python
# create a new column age-group
labels=['Young Adult','Adult','Middle-Age','Senior']
df['age group']=pd.qcut(df['age'],q=4,labels=labels)
```

```python
df[['age','age group']].head(10)
```

```python
# create column purchased_frequency day
frequancy_mapping={
    'Fortnightly':14,  
    'Weekly':7,
    'Annually':365,
    'Quarterly':90,
    'Bi-Weekly':14,
    'Monthly':30,
    'Every 3 Months':90
}
df['purchased frequency days']=df['frequency of purchases'].map(frequancy_mapping)
```

```python
df[['frequency of purchases','purchased frequency days']].head(10)
```

```python
df[['discount applied','promo code used']].head() 
```
